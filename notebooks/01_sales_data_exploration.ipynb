```python
financial_columns = ['revenue', 'total_cost', 'profit']

df[financial_columns].describe().T
```

```python
for col in financial_columns:
    Q1 = df[col].quantile(0.25)
    Q3 = df[col].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = df[
        (df[col] < lower_bound) |
        (df[col] > upper_bound)
    ]

    print(f"\n{col}")
    print("Lower bound:", lower_bound)
    print("Upper bound:", upper_bound)
    print("Potential outliers:", len(outliers))
```

### Order Structure Investigation

An investigation was conducted to understand the repeated order numbers and determine whether the records represent identical transactions.

The following findings were established:

- The dataset contains 10,684 unique order numbers.
- Every order number appears exactly six times.
- Customer, product, channel, quantity, unit price, revenue, total cost and profit remain consistent within each order.
- Dates and geographical locations vary within each order.
- Most orders are associated with multiple states and geographical regions.
- Almost all orders have six distinct dates and coordinates.
- Consecutive records within an order have irregular date intervals, with a mean interval of approximately 218 days.
- A total of 101 orders contain consecutive records on the same date. All have different geographical coordinates.
- Budget information is available for 8,616 orders (80.64%), while 2,068 orders (19.36%) have no available budget value.
- No order contains more than one distinct available budget value.
- All available budget records occur in 2017.

**Conclusion:**

The dataset has an unusual order structure. Each order number is associated with six records that share identical customer, product and financial information but differ in dates and geographical locations.

The business meaning of these repeated records has not been established. They may represent different geographical events or another reporting structure, but the available columns do not confirm this.

No records will be removed based solely on repeated order numbers. The dataset's intended level of detail must be clarified before making deduplication decisions or interpreting aggregate sales figures.
